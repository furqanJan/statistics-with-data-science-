# PROGRAMMING LOGIC & CONDITIONS


 Programming logic means thinking about a problem
 and deciding what the computer should do.

 Before writing code, we should ask:
1. What is the problem?

2. What information do we have?

3. What decision do we need to make?

4. What should happen if the condition is True?

#5. What should happen if the condition is False?

````python
# ============================================================
# ARITHMETIC OPERATORS
# ============================================================

# +  Addition
# -  Subtraction
# *  Multiplication
# /  Division
# %  Remainder
# ** Power

a = 10
b = 3

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a % b)
print(a ** b)
````

````python
# ============================================================
# COMPARISON OPERATORS
# ============================================================

# ==  Equal to
# !=  Not equal to
# >   Greater than
# <   Less than
# >=  Greater than or equal to
# <=  Less than or equal to

a = 10
b = 5

print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
````

````python
# ============================================================
# TRUE AND FALSE
# ============================================================

# Comparison operators give us either:
# True  -> The condition is correct
# False -> The condition is not correct

age = 20

print(age >= 18)
print(age < 18)
````

````python
# ============================================================
# IF STATEMENT
# ============================================================

# The if statement is used to make a decision.
#
# If the condition is True,
# Python will execute the code inside the if block.

age = 20

if age >= 18:
    print("You are an adult.")
````

````python
# ============================================================
# IF / ELSE
# ============================================================

# if    -> runs when the condition is True
# else  -> runs when the condition is False

age = 16

if age >= 18:
    print("You are an adult.")
else:
    print("You are under 18.")
````

````python
# ============================================================
# IF / ELIF / ELSE
# ============================================================

# elif means "else if".
# It allows us to check multiple conditions.

marks = 40

if marks >= 80:
    print("Grade A")
elif marks >= 60:
    print("Grade B")
elif marks >= 50:
    print("Grade C")
else:
    print("Fail")
````

````python
# ============================================================
# PROBLEM-SOLVING
# ============================================================

# Problem:
# A student needs at least 50 marks to pass.
#
# Step 1: We have the student's marks.
# Step 2: We have the passing marks.
# Step 3: Compare the two values.
# Step 4: Display Pass or Fail.

marks = 50
passing_marks = 50

if marks >= passing_marks:
    print("Student has passed.")
else:
    print("Student has failed.")
````

````python
# ============================================================
# PRACTICE PROBLEM
# ============================================================

# Ask the user to enter their marks.
# If marks are 50 or above -> Pass
# Otherwise -> Fail

marks = float(input("Enter your marks: "))

if marks >= 50:
    print("Pass")
else:
    print("Fail")
````

for Loop

````python
# ============================================================
# FOR LOOP
# ============================================================

# A for loop is used to repeat something
# for each item in a collection or sequence.

for i in range(5):
    print(i)
````

Simple for Loop

````python
# Print a message 5 times

for i in range(5):
    print("Hello Students")
````

while Loop

````python
# ============================================================
# WHILE LOOP
# ============================================================

# A while loop repeats the code
# as long as the condition is True.

number = 1

while number <= 5:
    print(number)
    number = number + 1
````

| `for` loop                                                        | `while` loop                                                       |
| ----------------------------------------------------------------- | ------------------------------------------------------------------ |
| Used when we know **how many times** we want to repeat something. | Used when we repeat something **until a condition becomes False**. |
| Works well with lists, strings, and `range()`.                    | Works mainly with a condition.                                     |
| The number of repetitions is usually known.                       | The number of repetitions may not be known.                        |


````python
# Repeat 5 times

for i in range(5):
    print("Hello")
````

````python
# Keep repeating while the number is less than 5

number = 1

while number < 5:
    print(number)
    number = number + 1
````


5. FUNCTIONS & REUSABLE CODE

 A function is a block of code that performs a specific task.

 We write a function once and can use it many times.

Functions help us:
 Avoid repeating code
 Organize our program
 Make code easier to understand

````python
# Creating a simple function

def hello():
    print("Hello Students!")

# Calling the function
hello()
````

````python
# A parameter allows us to give information to a function.

def hello(name):
    print("Hello", name)

hello("Furqan")
hello("Ali")
hello("Sara")
````

````python
# A function can have more than one parameter.

def add(a, b):
    print(a + b)

add(10, 5)
add(20, 30)
````

````python
# return sends the result back from the function.
#
# We can store the returned result in a variable.

def add(a, b):
    return a + b

result = add(10, 5)

print(result)
````

````python
# With return, we can use the result later.

def add(a, b):
    return a + b

result = add(10, 5)

print("Result:", result)
print("Result multiplied by 2:", result * 2)
````

````python
import tkinter as tk

# Create window
window = tk.Tk()
window.title("Simple Calculator")
window.geometry("300x300")

# First number
tk.Label(window, text="Enter first number").pack()
a = tk.Entry(window)
a.pack()

# Second number
tk.Label(window, text="Enter second number").pack()
b = tk.Entry(window)
b.pack()

# Calculate function
def calculate():
    num1 = float(a.get())
    num2 = float(b.get())

    addition = num1 + num2
    subtraction = num1 - num2
    multiplication = num1 * num2
    division = num1 / num2

    result.config(
        text=f"Addition: {addition}\n"
             f"Subtraction: {subtraction}\n"
             f"Multiplication: {multiplication}\n"
             f"Division: {division}"
    )

# Button
tk.Button(window, text="Calculate", command=calculate).pack(pady=20)

# Result
result = tk.Label(window, text="Result")
result.pack()

# Keep window open
window.mainloop()
````

### Definitions

**1D Array:**

A **one-dimensional array** is a collection of values arranged in a **single line**.

Example:

```python
[10, 20, 30, 40]
```

**2D Array:**

A **two-dimensional array** is a collection of values arranged in **rows and columns**, like a table.

Example:

```python
[
    [10, 20],
    [30, 40]
]
```


**3D Array:**

A **three-dimensional array** is a collection of **multiple 2D arrays**, like multiple tables stacked together.

Example:

```python
[
    [[10, 20],
     [30, 40]],

    [[50, 60],
     [70, 80]]
]
```

## 1. Line Graph

A line graph is used to show change or a trend.

For example:
- Temperature during a week

Easy to remember:

Line Graph = Change / Trend

````python
# Line Graph

days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]
temperature = [25, 27, 26, 30, 29]

plt.plot(days, temperature)

plt.title("Temperature During the Week")
plt.xlabel("Days")
plt.ylabel("Temperature")

plt.show()
````

## 3. Scatter Plot

A scatter plot is used to show the relationship between two numerical variables.

For example:

Hours Studied → Marks

We can use a scatter plot to see whether students who study more hours tend to get higher marks.

Easy to remember:

Scatter Plot = Relationship

````python
# Scatter Plot

hours = [1, 2, 3, 4, 5, 6]
marks = [50, 55, 60, 70, 80, 90]

plt.scatter(hours, marks)

plt.title("Study Hours vs Marks")
plt.xlabel("Hours Studied")
plt.ylabel("Marks")

plt.show()
````

### Understanding the Code

plt.scatter() creates a scatter plot.

Each point represents one observation.

For example:

1 hour → 50 marks

2 hours → 55 marks

3 hours → 60 marks

Scatter plots are very useful in Statistics and Machine Learning

because they help us understand relationships between variables.

## 4. Pie Chart

A pie chart shows how different parts make up a whole.

For example, how a student spends 24 hours in a day.

Easy to remember:

Pie Chart = Parts of a Whole

````python
# Pie Chart

activities = ["Study", "Sleep", "Games", "Other"]
hours = [6, 8, 3, 7]

plt.pie(hours, labels=activities, autopct="%1.1f%%")

plt.title("Daily Activities")

plt.show()
````

### Understanding the Code

plt.pie() creates a pie chart.

labels gives names to the different parts.

autopct displays the percentage.

The total of all parts should represent one whole.

For example:

Study = 6 hours

Sleep = 8 hours

Games = 3 hours

Other = 7 hours

Total = 24 hours

## 5. Histogram

A histogram is used to understand the distribution of numerical data.

For example, we can use a histogram to see how student marks are distributed.

It groups values into ranges called bins.

Easy to remember:

Histogram = Distribution

````python
# Histogram

marks = [45, 50, 55, 60, 62, 65, 70, 72, 75, 78, 80, 85, 90, 92, 95]

plt.hist(marks, bins=5)

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
````
